## 1. Import libraries

## 2. Load your dataset

```python
df = pd.read_csv("Student_08_dataset.csv")
display(df)
```

```python
print(list(df.shape))
print(list(df.columns))
df.info()
print(df.isnull().sum())
print(df.duplicated().sum())
```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
score_columns = ["Aptitude", "Reasoning", "Verbal", "Technical"]
scores = df[score_columns].to_numpy()
print("Average:", np.mean(scores))
print("Highest:", np.max(scores))
print("Lowest:", np.min(scores))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
df["Total"] = df["Aptitude"] + df["Reasoning"] + df["Verbal"] + df["Technical"]

# Calculate average marks
df["Average"] = df["Total"] / 4

# Display the result

print(df)
result = df[df["Average"] >= 70]

print(result)

result = df.sort_values(by="Average", ascending=False)

print(result)

print(result.head(1))

```

## 6. Your individual project challenge
**Challenge:** Calculate overall average; apply a minimum 50 in every section rule; list eligible candidates.

Write your solution below and explain what the result means.


```python
  # Calculate average of four sections
df["Overall_Average"] = (df["Aptitude"] + df["Reasoning"] + df["Verbal"] + df["Technical"]) / 4

print(df[["Name", "Overall_Average"]])


# Check if student scored 50 or more in every section
df["Eligible"] = (df["Aptitude"] >= 50) & (df["Reasoning"] >= 50) & (df["Verbal"] >= 50) & (df["Technical"] >= 50)

# Display eligible students
eligible_students = df[df["Eligible"] == True]

print("Eligible Candidates:")
print(eligible_students[["Student_ID", "Name", "Overall_Average"]])


print("Total Eligible:", len(eligible_students))
print("Not Eligible:", len(df) - len(eligible_students))
```
